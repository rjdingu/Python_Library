```python
# Univariate : Numerical


data = {'Salary': [25000, 30000, 28000, 35000, 40000, 27000, 32000, 38000, 36000, 29000]}
```

```python
df = pd.DataFrame(data, index = range(1,len(data['Salary'])+1))
df
```

```python
# Line :
plt.plot(df['Salary'],color = 'b',marker = 'D',linestyle =  ':')
plt.xlabel('Employee number')
plt.ylabel('Salary')
plt.title('Salary Distribution')
plt.grid()
plt.show()
```

```python
# Histogram
plt.hist(df['Salary'],bins = 5)
plt.show()
```

```python
# Boxplot
df.loc[11] = 5043
plt.boxplot(df['Salary'])
plt.show()
```

```python
df.drop(index = 11,inplace = True)
```

```python
df['Dept'] = ['IT', 'HR', 'FINANCE', 'HR', 'IT', 'FINANCE', 'IT', 'HR', 'FINANCE', 'IT']
df
```

```python
# Univeriate : Categorical
# Pie chart

count = df['Dept'].value_counts()
count
```

```python
plt.pie(count,labels = count.index, autopct = '%0.1f',explode = [0,0,0.2])
plt.show()
```

```python
# Bar Chart

plt.bar(count.index,count,color = ['blue','red','green'])
plt.show()
```

```python
df['Age'] = [25,30,28,35,40,27,32,38,36,29]
df
```

```python
# Scatter plot
plt.scatter(df['Salary'],df['Age'],marker = '^', color = 'orange')
plt.show()
```

```python
# Line Plot
plt.plot(df['Salary'],df['Age'],marker = '*',linestyle = '--',linewidth = 0.5)
plt
```

```python
# Bar Cart

plt.bar(df['Age'],df['Salary'], align = 'edge')

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Salary Chart')
plt.show()
```

```python
# Bivariate - dept : Salary

hr_sal = df[df['Dept'] == 'HR']['Salary']
it_sal = df[df['Dept'] == 'IT']['Salary']
fin_sal = df[df['Dept'] == 'FINANCE']['Salary']


hr_sal
```

```python
it_sal
```

```python
fin_sal
```

```python
# Box Plot
plt.boxplot([hr_sal,it_sal,fin_sal],labels = ['HR','IT','FINANCE'])
plt.show()
```

```python
salary_by_dept = df.groupby('Dept')['Salary'].sum()
salary_by_dept
```

```python
# Pie chart
plt.pie(salary_by_dept,labels = salary_by_dept.index,autopct = '%0.1f',explode = [0,0,0.2])
plt.show()
```
