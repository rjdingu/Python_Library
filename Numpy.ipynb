# Plotting Trignometric Curves

```python
x_sin = np.arange(0,2*pie,1)
y_sin = np.sin(x_sin)
print(y_sin)
```

```python
plt.figure(figsize = (6,6))
plt.plot(x_sin,y_sin)
plt.show()
```

```python
# for tan

x_tan =  np.arange(0,pie*2,1)
y_tan =  np.tan(x_tan)
print(y_tan)
```

```python
plt.figure(figsize = (3,3))
plt.plot(x_tan,y_tan,'h-')

plt.show()
```
